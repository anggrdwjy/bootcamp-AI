```python
#jika temp lebih dari sama dengan batas threshold maka panas
#jika tidak maka dingin lor

temp = 20 #ganti variable disini
threshold = 28

if temp >= threshold: #identasi 1
  print('panas') #identasi 2
elif temp <= threshold: #identasi 1
  print('sejuk')
else:
  print('dingin')
```

```python
#func kumpulan code yang dijadikan 1

def hitung_suhu():
  if temp >= threshold: #identasi 1
    print('panas') #identasi 2
  elif temp <= threshold: #identasi 1
    print('sejuk')
  else:
    print('dingin')
```

```python
#panggil func
hitung_suhu()
```

```python
#parameter
def hitung_suhu2(temp_2, threshold_s):

  if temp_2 >= threshold_s: #identasi 1
    print('panas') #identasi 2
  elif temp_2 <= threshold_s: #identasi 1
    print('sejuk')
  else:
    print('dingin')

  return temp_2 #untuk lanjutan
```

```python
kalkulasi_aja = hitung_suhu2(temp_2 = 2, threshold_s = 10)
```

```python
kalkulasi_aja + 1
```

```python
#mini challanges - 1

#data input
temps = [23, 25, 26, 29, 30] #list data
thershold = 29

#function
def weathermap_aja(temps, thershold):

  for temp in temps: #looping
    print(temp)

    if temp >= thershold:
      print('panas')
    elif temp <= thershold:
      print('dingin')
    else:
      print('sejuk')
```

```python
weathermap_aja(temps, thershold)
```
